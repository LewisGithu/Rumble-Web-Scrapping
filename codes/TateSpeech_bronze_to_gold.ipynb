```sql
-- Classify each Emergency Meeting episode by its title.
-- The AI result is sampled, so the table is built from ONE call per title.
-- Manually reviewed overrides (below) deterministically win over the AI label.
CREATE OR REPLACE TABLE projects.rumble.emergency_meetings_classified AS
WITH classified AS (
  SELECT
    e.*,
    ai_classify(
      Title,
      '{
        "Culture": "Society, art, entertainment, or social commentary",
        "Lifestyle": "Everyday life, travel, relationships, cars, or personal enjoyment",
        "Crypto": "Cryptocurrency, Bitcoin, blockchain, or digital money",
        "Finance": "Stock markets, investing, business, or the economy",
        "Money": "General money, wealth, getting rich, or income topics that are not about markets, investing, or crypto",
        "Mindset": "Motivation, discipline, self-improvement, mental attitude, or success philosophy",
        "Fitness": "Exercise, training, bodybuilding, health, or physical performance",
        "Grinding": "Working hard, hustling, relentless effort, struggle, or being in the trenches",
        "Court Case": "Legal proceedings, trials, court rulings, lawyers, lawsuits, or case updates",
        "Interviews": "A guest interview, debate, or conversation with someone",
        "Matrix": "Fighting the establishment, the system, media, censorship, government overreach, or institutional power",
        "Politics": "Politicians, elections, government policy, war, or international conflict",
        "Other": "The title gives no clear topic signal"
      }',
      MAP(
        'version', '2.1',
        'instructions', 'Inputs are short titles of episodes from a talk show called Emergency Meeting. Pick the single best-fit topic from the words in the title; prefer the most specific label (e.g. Fitness over Lifestyle, Finance or Crypto over Money); if there is no clear signal, pick other.',
        'enableConfidenceScores', 'true'
      )
    ) AS classification
  FROM projects.rumble.emergency_meetings e
),
ai_topics AS (
  SELECT
    classified.*,
    classification:response[0].value::STRING AS ai_topic,
    classification:response[0].confidence_score::DOUBLE AS ai_confidence
  FROM classified
),
-- Manual topic overrides reviewed by the user: matched by title pattern,
-- they deterministically win over the sampled AI label.
overrides AS (
  SELECT '%HARD TO KILL%' AS pattern, 'Mindset' AS topic
  UNION ALL SELECT '%THREE LIVES%', 'Matrix'
  UNION ALL SELECT '%HUGE CASE UPDATE%', 'Court Case'
  UNION ALL SELECT '%SO BACK%', 'Politics'
  UNION ALL SELECT '%CASE DISMISSED%', 'Court Case'
  UNION ALL SELECT '%IN THE TRENCHES%', 'Grinding'
  UNION ALL SELECT '%100K PARTY%', 'Politics'
  UNION ALL SELECT '%TRUMP PUMP%', 'Crypto'
  UNION ALL SELECT '%THE RETURN%', 'Politics'
  UNION ALL SELECT '%CONTROVERSY%', 'Matrix'
  UNION ALL SELECT '%HQ EXPLAINED%', 'Lifestyle'
  UNION ALL SELECT '%HOW THE WORLD WORKS%', 'Matrix'
  UNION ALL SELECT '%THE TRUTH IS OUT%', 'Matrix'
  UNION ALL SELECT '%SLANDEROUS LIES%', 'Matrix'
  UNION ALL SELECT '%SWORD%', 'Matrix'
  UNION ALL SELECT '%MERRY CHRISTMAS%', 'Culture'
  UNION ALL SELECT '%GRAND ADVENTURE%', 'Lifestyle'
)
SELECT
  a.* EXCEPT (ai_topic, ai_confidence),
  COALESCE(o.topic, a.ai_topic) AS topic,
  CASE WHEN o.topic IS NOT NULL THEN 1.0 ELSE a.ai_confidence END AS confidence
FROM ai_topics a
LEFT JOIN overrides o
  ON upper(a.Title) LIKE o.pattern;
```

```sql
-- Inspect table history so we can diff the previous classification against the current one
DESCRIBE HISTORY projects.rumble.emergency_meetings_classified;
```

```sql
-- Diff previous classification (version 2) vs current (version 3): which labels changed?
WITH prev AS (
  SELECT Title, topic FROM projects.rumble.emergency_meetings_classified VERSION AS OF 2
),
curr AS (
  SELECT Title, topic FROM projects.rumble.emergency_meetings_classified VERSION AS OF 3
)
SELECT prev.topic AS previous_topic, curr.topic AS current_topic, COUNT(*) AS n
FROM prev
JOIN curr USING (Title)
WHERE prev.topic <> curr.topic
GROUP BY 1, 2
ORDER BY n DESC, 1, 2;
```

```sql
-- List the episodes that were previously labeled but flipped to Other in the latest run
WITH prev AS (
  SELECT Title, topic, confidence FROM projects.rumble.emergency_meetings_classified VERSION AS OF 2
),
curr AS (
  SELECT Title, Date, topic, confidence FROM projects.rumble.emergency_meetings_classified VERSION AS OF 3
)
SELECT curr.Title, prev.topic AS previous_topic, prev.confidence AS previous_confidence
FROM prev
JOIN curr USING (Title)
WHERE prev.topic <> 'Other'
  AND curr.topic = 'Other'
ORDER BY prev.topic, curr.Date DESC;
```

```sql
-- Restore previously confirmed labels with a direct UPDATE (single atomic statement).
-- No AI re-sampling happens, so no other episode's label can drift.
UPDATE projects.rumble.emergency_meetings_classified
SET
  topic = CASE Title
    WHEN 'EMERGENCY MEETING EPISODE 120 - HOW THE WORLD WORKS' THEN 'Matrix'
    WHEN 'EMERGENCY MEETING EPISODE 116 - THE TRUTH IS OUT'     THEN 'Matrix'
    WHEN 'EMERGENCY MEETING EPISODE 66 - SLANDEROUS LIES'        THEN 'Matrix'
    WHEN 'EMERGENCY MEETING EPISODE 52 - DO YOU HAVE A SWORD?'   THEN 'Matrix'
    WHEN 'EMERGENCY MEETING EPISODE 29 - MERRY CHRISTMAS'        THEN 'Culture'
    WHEN 'Emergency Meeting Episode 22 - The Grand Adventure'     THEN 'Lifestyle'
    ELSE topic
  END,
  confidence = CASE
    WHEN Title IN (
      'EMERGENCY MEETING EPISODE 120 - HOW THE WORLD WORKS',
      'EMERGENCY MEETING EPISODE 116 - THE TRUTH IS OUT',
      'EMERGENCY MEETING EPISODE 66 - SLANDEROUS LIES',
      'EMERGENCY MEETING EPISODE 52 - DO YOU HAVE A SWORD?',
      'EMERGENCY MEETING EPISODE 29 - MERRY CHRISTMAS',
      'Emergency Meeting Episode 22 - The Grand Adventure'
    ) THEN 1.0
    ELSE confidence
  END
WHERE Title IN (
  'EMERGENCY MEETING EPISODE 120 - HOW THE WORLD WORKS',
  'EMERGENCY MEETING EPISODE 116 - THE TRUTH IS OUT',
  'EMERGENCY MEETING EPISODE 66 - SLANDEROUS LIES',
  'EMERGENCY MEETING EPISODE 52 - DO YOU HAVE A SWORD?',
  'EMERGENCY MEETING EPISODE 29 - MERRY CHRISTMAS',
  'Emergency Meeting Episode 22 - The Grand Adventure'
);
```

```sql
-- Add a clean episode title: strip the "Emergency Meeting [Episode X] - " prefix.
-- The original Title stays untouched (overrides and diffs match on it).
ALTER TABLE projects.rumble.emergency_meetings_classified ADD COLUMN episode_title STRING;

UPDATE projects.rumble.emergency_meetings_classified
SET episode_title = regexp_replace(Title, '(?i)^\\s*EMERGENCY\\s+MEETING(\\s+(EPISODE\\s+)?\\d+)?\\s*-\\s*', '');

-- Verify: every row has a non-empty episode_title, and prefixes were actually stripped
SELECT
  COUNT(*) AS total_rows,
  SUM(CASE WHEN episode_title IS NULL OR episode_title = '' THEN 1 ELSE 0 END) AS missing_title,
  SUM(CASE WHEN episode_title = Title THEN 1 ELSE 0 END) AS not_trimmed
FROM projects.rumble.emergency_meetings_classified;
```

```sql
CREATE OR REPLACE TABLE projects.rumble.gold_t_table AS
SELECT
  Date,
  Time,
  episode_title AS Title,
  -- Duration arrives as either "H:MM:SS" (e.g. 1:03:26) or "MM:SS" (e.g. 45:07);
  -- convert both to total minutes.
  ROUND(
    CASE WHEN array_size(split(Duration, ':')) = 3
      THEN CAST(split_part(Duration, ':', 1) AS DOUBLE) * 60.0
         + CAST(split_part(Duration, ':', 2) AS DOUBLE)
         + CAST(split_part(Duration, ':', 3) AS DOUBLE) / 60.0
      ELSE CAST(split_part(Duration, ':', 1) AS DOUBLE)
         + CAST(split_part(Duration, ':', 2) AS DOUBLE) / 60.0
    END,
  2) AS Duration_minutes,
  topic AS Topic,
  Views,
  Likes,
  Comments
FROM projects.rumble.emergency_meetings_classified;

-- Verify: spot-check the conversion against the raw Duration string
SELECT g.Title, c.Duration AS raw_duration, g.Duration_minutes
FROM projects.rumble.gold_t_table g
JOIN projects.rumble.emergency_meetings_classified c
  ON g.Title = c.episode_title
ORDER BY g.Date DESC;
```
